```python
#Task2
def analyze_file(filename):
    total_lines = 0
    invalid_lines = 0

    with open(filename, "r") as file:
        for line_number, line in enumerate(file, start=1):
            total_lines += 1
            line = line.strip()

            if not line:  # Skip empty lines
                continue
            
            parts = line.split(',')
            if len(parts) != 26:
                print(f"Invalid line {line_number}: does not contain exactly 26 values:\n{line}")
                invalid_lines += 1
                continue
            
            student_id = parts[0]
            if not (student_id.startswith('N') and len(student_id) == 9 and student_id[1:].isdigit()):
                print(f"Invalid line {line_number}: N# is invalid\n{line}")
                invalid_lines += 1
                continue

            # Check for valid answers (A, B, C, D) or empty responses
            valid_answers = {'A', 'B', 'C', 'D', ''}
            answers = parts[1:]  # Get the answers

            for answer in answers:
                if answer not in valid_answers:
                    print(f"Invalid answer on line {line_number}: {answer} is not a valid response\n{line}")
                    invalid_lines += 1
                    break  # No need to check further answers if one is invalid

    return total_lines, invalid_lines

def main():
    while True:
        try:
            filename = input("Enter a class file to grade (i.e. class1 for class1.txt): ")
            if not filename.endswith(".txt"):
                filename += ".txt"
            
            with open(filename, "r") as file:
                print(f"Successfully opened {filename}")
                
                print("**** ANALYZING ****")
                total_lines, invalid_lines = analyze_file(filename)
                
                if invalid_lines == 0:
                    print("No errors found!")
                
                print("**** REPORT ****")
                print(f"Total valid lines of data: {total_lines - invalid_lines}")
                print(f"Total invalid lines of data: {invalid_lines}")
                break
                
        except FileNotFoundError:
            print("File cannot be found. Please try again.")
        except Exception as e:
            print(f"An error occurred: {e}")

if __name__ == "__main__":
    main()
```

```python
#task3
def analyze_exam_file():
    # Đáp án đúng
    answer_key = "B,A,D,D,C,B,D,A,C,C,D,B,A,B,A,C,B,D,A,C,A,A,B,D,D".split(',')
    
    # Nhập tên tệp
    filename = input("Enter a class to grade (i.e. class1 for class1.txt): ").strip() + ".txt"
    
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()
        print(f"Successfully opened {filename}")
    except FileNotFoundError:
        print(f"Error: File {filename} not found.")
        return

    print("**** ANALYZING ****")

    valid_data = []
    invalid_data = []

    for line in lines:
        line = line.strip()
        parts = line.split(',')

        # Kiểm tra định dạng ID và số lượng câu trả lời
        if len(parts) != 26 or not parts[0].startswith('N') or not parts[0][1:].isdigit() or len(parts[0]) != 9:
            invalid_data.append(line)
        else:
            valid_data.append(parts)

    # Báo cáo lỗi
    for line in invalid_data:
        if len(line.split(',')) != 26:
            print(f"Invalid line of data: does not contain exactly 26 values:\n{line}")
        elif not line.split(',')[0].startswith('N') or not line.split(',')[0][1:].isdigit():
            print(f"Invalid line of data: N# is invalid\n{line}")

    print("**** REPORT ****")
    print(f"Total valid lines of data: {len(valid_data)}")
    print(f"Total invalid lines of data: {len(invalid_data)}")

    if not valid_data:
        print("No valid data to process.")
        return

    # Chấm điểm
    scores = []
    skipped = [0] * 25
    incorrect = [0] * 25

    for student in valid_data:
        score = 0
        for i, answer in enumerate(student[1:]):
            if answer == '':
                skipped[i] += 1
            elif answer == answer_key[i]:
                score += 4
            else:
                score -= 1
                incorrect[i] += 1
        scores.append(score)

    # Thống kê điểm số
    high_scores = sum(1 for s in scores if s > 80)
    mean_score = round(sum(scores) / len(scores), 3)
    highest_score = max(scores)
    lowest_score = min(scores)
    score_range = highest_score - lowest_score
    
    sorted_scores = sorted(scores)
    if len(sorted_scores) % 2 == 1:
        median_score = sorted_scores[len(sorted_scores) // 2]
    else:
        mid = len(sorted_scores) // 2
        median_score = round((sorted_scores[mid - 1] + sorted_scores[mid]) / 2, 3)

    # Phân tích câu hỏi bị bỏ qua nhiều nhất
    max_skipped = max(skipped)
    most_skipped = [(i + 1, skipped[i], round(skipped[i] / len(valid_data), 3)) for i in range(25) if skipped[i] == max_skipped]

    # Phân tích câu hỏi bị trả lời sai nhiều nhất
    max_incorrect = max(incorrect)
    most_incorrect = [(i + 1, incorrect[i], round(incorrect[i] / len(valid_data), 3)) for i in range(25) if incorrect[i] == max_incorrect]

    # Báo cáo kết quả
    print(f"\nTotal student of high scores: {high_scores}")
    print(f"Mean (average) score: {mean_score}")
    print(f"Highest score: {highest_score}")
    print(f"Lowest score: {lowest_score}")
    print(f"Range of scores: {score_range}")
    print(f"Median score: {median_score}")

    print("\nQuestion that most people skip:")
    for q in most_skipped:
        print(f"{q[0]} - {q[1]} - {q[2]}")

    print("\nQuestion that most people answer incorrectly:")
    for q in most_incorrect:
        print(f"{q[0]} - {q[1]} - {q[2]}")

if __name__ == "__main__":
    analyze_exam_file()
```

```python
#Task4
import os

def save_grades_to_file(filename, student_scores):
    # Lấy tên gốc của tệp
    base_name = os.path.splitext(filename)[0]
    # Tạo tên tệp mới để lưu điểm
    output_filename = f"{base_name}_grades.txt"
    
    # Ghi dữ liệu vào tệp
    with open(output_filename, "w") as file:
        for student_id, score in student_scores.items():
            file.write(f"{student_id},{score}\n")
    
    print(f"Grades have been saved to {output_filename}")
    
    # Đọc và in chỉ student ID và điểm từ tệp đã lưu
    read_and_print_file(output_filename)

def read_and_print_file(filename):
    try:
        with open(filename, "r") as file:
            print("\nStudent ID and Score:")
            for line in file:
                print(line.strip())  # In từng dòng, chỉ chứa student ID và điểm
    except FileNotFoundError:
        print(f"File {filename} not found!")

def analyze_exam_file():
    # Đáp án đúng
    answer_key = "B,A,D,D,C,B,D,A,C,C,D,B,A,B,A,C,B,D,A,C,A,A,B,D,D".split(',')
    
    # Nhập tên tệp
    filename = input("Enter a class to grade (i.e. class1 for class1.txt): ").strip() + ".txt"
    
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()
        print(f"Successfully opened {filename}")
    except FileNotFoundError:
        print(f"Error: File {filename} not found.")
        return

    valid_data = []

    for line in lines:
        line = line.strip()
        parts = line.split(',')

        # Kiểm tra định dạng ID và số lượng câu trả lời
        if len(parts) == 26 and parts[0].startswith('N') and parts[0][1:].isdigit() and len(parts[0]) == 9:
            valid_data.append(parts)

    if not valid_data:
        print("No valid data to process.")
        return

    # Chấm điểm
    student_scores = {}

    for student in valid_data:
        score = 0
        student_id = student[0]
        for i, answer in enumerate(student[1:]):
            if answer == '':
                continue  # Skip blank answers
            elif answer == answer_key[i]:
                score += 4
            else:
                score -= 1
        student_scores[student_id] = score

    # Save grades to file
    save_grades_to_file(filename, student_scores)

if __name__ == "__main__":
    analyze_exam_file()
```
